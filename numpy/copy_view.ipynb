```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5])
copy = arr.copy()
view = arr.view()
```

```python
# Change the array
arr[0] = 10
print('After changing array : ')
print(f'Array {arr}')
print(f'View {view}')
print(f'Copy {copy}')
```

```python
# Change copy
copy[1] = 20
print('After changing copy : ')
print(f'Array {arr}')
print(f'View {view}')
print(f'Copy {copy}')
```

```python
# Change view
view[2] = 30
print('After changing view : ')
print(f'Array {arr}')
print(f'View {view}')
print(f'Copy {copy}')
```
