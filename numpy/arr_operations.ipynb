```python
import numpy as np
```

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
```

```python
# Arithmetic operations

# Add (size of both array must be same)
print(arr1 + arr2)
```

```python
# Subtract (size of both array must be same)
print(arr2 - arr1)
```

```python
# Multiply (size of both array must be same)
print(arr1 * arr2)
```

```python
# Divide (size of both array must be same)
print(arr2 / arr1)
```

```python
# Floor division (size of both array must be same)
print(arr2 // arr1)
```

```python
# Exponent (size of both array must be same)
print(arr2 ** arr1)
```

```python
# Broadcasting
print(arr1 + 10) # Add 10 to all elements of arr1
print(arr2 - 5) # Subtract 5 from all elements of arr2
```

```python
# Broadcasting in 2-d array
arr3 = np.arange(1,21).reshape(4,5)
print(arr3)
print(arr3 + 10)
```

```python
# Deep and Shallow copy
a = np.arange(1,11)
print(a)
shallow = a
print(shallow)
deep = a.copy()
print(deep)
slice = a[:6]
print(slice)
```

```python
a[0] = 11
print(a)
print(shallow)
print(deep)
print(slice) # in slice we have 11 , because in numpy slicing returns view, not a copy
```

```python
slice[2] = 200 # this will give 200 in 2 index in shallow and a both because slice is view so change in slice will reflect in original array
a[1] = 100
print(a)
print(shallow)
print(deep)
print(slice)
```
