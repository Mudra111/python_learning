```python
import numpy as np
```

```python
# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
print(data)
```

```python
# Shape of the matrix
data.shape
```

```python
# Mean of age of all students
print(np.mean(data[:,0]))
```

```python
# Math marks of all students
print(data[:,1])
```

```python
#Find the highest Science mark
print(data[:,2].max())
# or
print(np.max(data[:,2]))
```

```python
#Get details of the student who scored more than 90 in Math
data[data[:,1] > 90]
```

```python
#Increase Math marks of all students by 5
data[:,1] += 5
data
```

```python
#Find how many students are younger than 19
len(data[data[:,0] < 19])
```

```python
#Calculate the average marks in each subject (column-wise mean)
np.mean(data[:,1:],0)
```

```python
#Get data of students who scored at least 80 in both subjects
print(data[(data[:,1] >= 80) & (data[:,2] >= 80)])
```

```python
#Replace all Science marks < 75 with 0
data[:,2][data[:,2] < 75] = 0 
data
```
