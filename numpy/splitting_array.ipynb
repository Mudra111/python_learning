```python
import numpy as np
```

```python
arr = np.arange(36).reshape(6,6)
print(arr)
```

```python
hsplit = np.hsplit(arr,2)
print(hsplit)
part1 = hsplit[0]
part2 = hsplit[1]
print(part1)
print(part2)
```

```python
vsplit = np.vsplit(arr,3)
print(vsplit)
vpart1 = vsplit[0]
vpart2 = vsplit[1]
vpart3 = vsplit[2]
print(vpart1)
print(vpart2)
print(vpart3)
```
