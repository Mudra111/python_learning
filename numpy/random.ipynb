```python
import numpy as np
```

```python
# Generate 10 random numbers between 0-1 (normalization)
random = np.random.rand(10)
print(random)
```

```python
# Generate 10 random numbers between -3 & 3 (standardization)
random1 = np.random.randn(10)
print(random1)
```

```python
# Generate 10 random digits (low,high,size)
random2 = np.random.randint(1,100,10)
print(random2)
```
