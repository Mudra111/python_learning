```python
import numpy as np
```

```python
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
print(arr)
print(len(arr))
```

```python
# find first quartile - Q1
Q1 = np.percentile(arr,25)
Q1
```

```python
# find third quartile - Q3
Q3 = np.percentile(arr,75)
Q3
```

```python
# find IQR - Interquartile range
IQR = Q3 - Q1
IQR
```

```python
# find uppen fence
upper_fence = Q3 + 1.5*IQR
upper_fence
```

```python
# find lower fence
lower_fence = Q1 - 1.5*IQR
lower_fence
```

```python
# remove outliers which are larger than upper_fence and lower than lower_fence
without_outlier = (arr > -20) & (arr < 52)
arr_new = arr[without_outlier]
arr_new
```

```python
# 5 point summary - after removing outliers
minimum = arr_new.min()
first_quartile = np.percentile(arr_new,25)
median = np.median(arr_new)
third_quartile = np.percentile(arr_new,75)
maximum = arr_new.max()
```

```python
print(f'Minimum : {minimum}')
print(f'First Quartile : {first_quartile}')
print(f'Median : {median}')
print(f'Third Quartile : {third_quartile}')
print(f'Maximum : {maximum}')
```
