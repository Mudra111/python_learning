```python
df_cleaned = df
df_cleaned
```

```python
df_cleaned.drop_duplicates(inplace=True)
df_cleaned
```

```python
df_cleaned['Sex'].value_counts()
```

```python
df_cleaned['Sex'] = df_cleaned['Sex'].map({'M':0,'F':1})
df_cleaned.rename(columns={'Sex':'is_female'},inplace=True)
```

```python
df_cleaned['is_female'].value_counts()
```

```python
df_cleaned['ExerciseAngina'] = df_cleaned['ExerciseAngina'].map({'N':0,'Y':1})
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['ChestPainType','RestingECG','ST_Slope'])
df_cleaned = df_cleaned.astype(int)
```

```python
df_cleaned.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
# col = ['Age','RestingBP','Cholesterol','MaxHR']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df_cleaned.drop('HeartDisease',axis=1)
y = df_cleaned['HeartDisease']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
```

```python
from sklearn.metrics import accuracy_score,f1_score,classification_report
```

```python
# create dictionary for all models
models = {
    'Logistic_regression' : LogisticRegression(),
    'Decision_tree' : DecisionTreeClassifier(),
    'Naive_bayes' : GaussianNB(),
    'Knn': KNeighborsClassifier(),
    'Svm' : SVC()
}
results = []
```

```python
# try all model and calculate accuracy and f1 score
for name,model in models.items():
    model.fit(X_train_scaled,y_train)
    y_pred = model.predict(X_test_scaled)
    accuracy = accuracy_score(y_test,y_pred)
    f1 = f1_score(y_test,y_pred)
    results.append({
        'model':name,
        'accuracy':round(accuracy,4),
        'f1_score':round(f1,4)
    })
```
